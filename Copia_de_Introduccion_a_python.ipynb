##Listas

Las listas son colecciones ordenadas y modificables. Permiten elementos duplicados.

"Las listas son versátiles y pueden ser modificadas, lo que las hace ideales para almacenar colecciones de elementos que pueden cambiar."

```python
# @title Ejemplo con  Listas
# Crear una lista
frutas = ["uva","manzana",  "banana", "cereza", "naranja", "sandia", True, False, 124, 1.232435]
print(len(frutas))
# Acceder a elementos
print(frutas[-1])  # Output: manzana
print("\n")
print(frutas[-10])  # Output: banana
print("\n")
# Modificar un elemento


frutas[-1] = "mango"
print(frutas)  # Output: ['manzana', 'mango', 'cereza']

print(type(frutas[-1]))

#print(frutas[-1]+frutas[0])
try:
    print(frutas[-1]+frutas[0])
    print(frutas[-1]+frutas[-2])

except Exception as e:
    # Guardar el error en una variable
    error_mensaje = str(e)
    print("Ocurrió un error:", error_mensaje)
```

```python
# @title Operaciones con Listas

# Añadir elementos

frutas.append("naranja")
print(frutas)  # Output: ['manzana', 'mango', 'cereza', 'naranja']

# Eliminar elementos
frutas.remove(frutas[0])
print(frutas)  # Output: ['manzana', 'cereza', 'naranja']

# Longitud de la lista
print(len(frutas))  # Output: 3

# Recorrer una lista
for fruta in frutas:
    print(fruta)



```

```python
# Lista original
mi_lista = [1, 2, 3, 4, 5]

# Mover el elemento en la posición 1 (valor 2) a la posición 3
elemento = mi_lista.pop(1)  # Extrae el elemento en la posición 1
#
mi_lista.insert(3, elemento)  # Inserta el elemento en la posición 3
mi_lista.append(elemento)
print(mi_lista)
```

##Tuplas
Las tuplas son colecciones ordenadas e inmutables. Permiten elementos duplicados.

"Las tuplas son inmutables, lo que significa que no puedes cambiar su contenido después de crearlas. Son útiles para almacenar datos constantes."

```python
# @title Ejemplo con Tuplas
# Crear una tupla
tupla = (5,9,3,5,7,2)
print(tupla)
# Acceder a elementos
print(tupla[0])  # Output: 1
print(tupla[1])  # Output: 2
colores = ("rojo", "verde", "azul", 12, 12.2443, True, False,"rojo", "verde", "azul", 12, 12.2443, True, False)
print(colores)
# Acceder a elementos
print(colores[0])  # Output: rojo
print(colores[-1])  # Output: verde

#colores[0] = "verde"
# Intentar modificar un elemento (esto causará un error)
# colores[1] = "amarillo"  # TypeError: 'tuple' object does not support item assignment
```

```python
# @title Operaciones con Tuplas
# Longitud de la tupla
print(len(colores))  # Output: 3

# Recorrer una tupla
for color in colores:
    print(color)

# Convertir una tupla a lista
lista_colores = list(colores)
print(lista_colores)  # Output: ['rojo', 'verde', 'azul']

lista_colores[0] = "cereza"
print (lista_colores)
# Convertir una lista a tupla
tupla_colores = tuple(lista_colores)
print(tupla_colores)  # Output: ('rojo', 'verde', 'azul')
```

##Diccionarios
Los diccionarios son colecciones desordenadas, modificables e indexadas. No permiten elementos duplicados.

Los elementos de un diccionario estan formados por una llave (key) y un valor (value)



```python
# @title Ejemplo con Diccionario
# Crear un diccionario
alumno = {"nombre": "Ana", "edad": 23, "carrera": "Ingeniería", "hobbie": True}

# Acceder a elementos
print(alumno["nombre"])  # Output: Ana
print(alumno["edad"])
   # Output: 23

# Modificar un valor
alumno["edad"] = "veinticuatro"
print(alumno)  # Output: {'nombre': 'Ana', 'edad': 24, 'carrera': 'Ingeniería'}
```

```python
# @title Operaciones con Diccionarios
alumno["materiras"]= {"historia": 9.6, "matematicas":6.7}
# Añadir elementos
alumno["promedio"] = 9.5
print(alumno)  # Output: {'nombre': 'Ana', 'edad': 24, 'carrera': 'Ingeniería', 'promedio': 9.5}

# Eliminar elementos
#del alumno["carrera"]
#print(alumno)  # Output: {'nombre': 'Ana', 'edad': 24, 'promedio': 9.5}
print(alumno["materiras"]["matematicas"])
# Longitud del diccionario
print(len(alumno))  # Output: 3

# Recorrer un diccionario
for key, value in alumno.items():
    print(key, value)

for clave in alumno:
    print(clave)
```

##Conjuntos (Sets)
Los conjuntos son colecciones desordenadas, no indexadas y no permiten elementos duplicados.

```python
# @title Ejemplo con sets
# Crear un conjunto
numeros = {1,1, 2, 3, 4, 5, 4.5, 1,"avión", "raton", "raton", "zopilote"}
print(numeros)
# Acceder a elementos (no se puede acceder por índice)
print(3 in numeros)  # Output: True
print(6 in numeros)  # Output: False
```

```python
# @title Operaciones con Conjuntos
# Añadir elementos
numeros.add(2)
print(numeros)  # Output: {1, 2, 3, 4, 5, 6}

# Eliminar elementos
numeros.remove(2)
print(numeros)  # Output: {1, 3, 4, 5, 6}

# Longitud del conjunto
print(len(numeros))  # Output: 5

# Recorrer un conjunto
for n in numeros:
    print(n)
```

#Funciones

Las funciones son lineas de codigo estrucutradas que permiten definir un bloque de código reutilizable que se puede ejecutar muchas veces dentro de un programa. Se trata de una estructura que agrupa partes del código creando soluciones modulares para problemas complejos.

Las funciones en Python cumplen objetivos específicos definidos por el usuario o por el lenguaje. Reciben como parámetros datos de entrada llamados argumentos, que se indica por el usuario o por defecto, los procesan y devuelven datos de salida.
Las funciones en Python sirven para:

* Dividir y ordenar el código en partes más sencillas para depurar y programar con mayor facilidad.
* Reutilizar el código, evitando repeticiones innecesarias en un programa.

Podemos distinguir dos tipos de funciones:

* Nativas (Built-in finctions): opciones disponibles que ya están integradas en Python.
* Personalizadas: creadas por el usuario. Para usarlas, necesitas definir las funciones que cumplan con las necesidades de tu proyecto.

```python
# @title Función Suma

def resta (a,b):
  c = a-b
  return c

def suma(a, b,c):
    return a + b +c

def division(a, b):
    return a / b

resultado = resta(1000000,2.5)
print (resultado)

print(suma(12,15,14))
print(division(10,2))

def area_circulo(r):
  return(3.1416*r*r)


print(area_circulo(5))
```

#Módulos

Si sales del intérprete de Python y vuelves a entrar, las definiciones que habías hecho (funciones y variables) se pierden. Por lo tanto, si quieres escribir un programa más o menos largo, es mejor que utilices un editor de texto para preparar la entrada para el intérprete y ejecutarlo con ese archivo como entrada. Esto se conoce como crear un script. A medida que tu programa crezca, quizás quieras separarlo en varios archivos para que el mantenimiento sea más sencillo. Quizás también quieras usar una función útil que has escrito en distintos programas sin copiar su definición en cada programa.

Para soportar esto, Python tiene una manera de poner definiciones en un archivo y usarlos en un script o en una instancia del intérprete. Este tipo de ficheros se llama módulo; las definiciones de un módulo pueden ser importadas a otros módulos o al módulo principal (la colección de variables a las que tienes acceso en un script ejecutado en el nivel superior y en el modo calculadora).

Para mayor información visiten la siguiente liga

https://docs.python.org/es/3/tutorial/modules.html

```python
import math

pi = math.pi

print(pi)


# Calcular la raíz cuadrada de un número
numero = 16
raiz_cuadrada = math.sqrt(numero)
print(f"La raíz cuadrada de {numero} es {raiz_cuadrada}")

# Calcular el seno de un ángulo (en radianes)
angulo = 3.1416 / 2  # 90 grados en radianes
seno = math.sin(angulo)
print(f"El seno de {angulo} radianes es {seno}")

# Calcular el logaritmo natural de un número
numero_log = 10
logaritmo = math.log(numero_log)
print(f"El logaritmo natural de {numero_log} es {logaritmo}")
```

```python
import math

def area_cilindro(radio, altura):
    # Calcular el área lateral
    area_lateral = 2 * math.pi * radio * altura

    # Calcular el área de las dos bases
    area_bases = 2 * math.pi * radio**2

    # Calcular el área total
    area_total = area_lateral + area_bases

    return area_total

# Ejemplo de uso
radio = float(input("Introduce el radio del cilindro: "))
altura = float(input("Introduce la altura del cilindro: "))

area = area_cilindro(radio, altura)
print(f"El área del cilindro es: {area}")
```
